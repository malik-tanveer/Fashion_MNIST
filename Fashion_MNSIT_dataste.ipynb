````python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.datasets import fashion_mnist
````

````python
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()
````

````python
print(train_images.shape)
print(test_images.shape)
print(train_labels.shape)
print(test_labels.shape)
````

# Data Normalization

## Train Images

Image pixel values are initially in the `0–255` range. This converts them to the `0–1` range, which can help the model during training.

```python
train_images = train_images.astype("float32") / 255.0
```

## Test Images
Test image pixel values are initially in the `0–255` range. This converts them to the `0–1` range, helping maintain consistency between training and testing data.

```python
test_images = test_images.astype("float32") / 255
```

````python
train_images = train_images.reshape((-1, 28 , 28, 1))
train_images = train_images.astype("float32") / 255
test_images = test_images.reshape((-1, 28 , 28,1 ))
test_images = test_images.astype("float32") / 255
````

# Model Squential define
1. `Conv2D` — Extracts features from clothing images.
2. `MaxPooling2D` — Reduces the size of feature maps.
3. `Flatten` — Converts feature maps into a one-dimensional vector.
4. `Dense(64)` — Processes the extracted features.
5. `Dense(8)` — Predicts one of the 8 clothing categories in Fashion-MNIST.

````python
model = keras.Sequential([
    layers.Input(shape=(28,28,1)),

    layers.Conv2D(8, kernel_size=3, activation="relu"),
    layers.MaxPool2D(pool_size=2),

    layers.Flatten(),

    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])
````

# Model Compile


1. `optimizer="adam"` — Updates the model's weights to improve its predictions during training.
2. `loss="sparse_categorical_crossentropy"` — Measures how far the model's predictions are from the correct categories. It is suitable for multi-class classification when labels are integer values, such as `0, 1, 2, ...`.
3. `metrics=["accuracy"]` — Tracks the percentage of predictions that are correct during training and evaluation.

````python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
````

### `model.fit()`

`model.fit()` trains the model using images and their correct labels.

- `train_images` — Images used for training.
- `train_labels` — Correct categories of those images.
- `epochs=5` — The model goes through the training data 5 times.
- `batch_size=32` — Processes 32 images per batch.
- `validation_split=0.1` — Reserves 10% of the training data for validation.

````python
model.fit(
    train_images,
    train_labels,
    epochs=5,
    batch_size=32,
    validation_split=0.1
)
````

# Evalute the CNN Model

````python
test_loss, test_accuracy = model.evaluate(test_images, test_labels)
````

````python
print("Test Accuracy:", test_accuracy)
````
